```python

df1=pd.read_parquet(r'..\data\training\vynfi\train-00000-of-00003.parquet')

print(df1.head())
print(df1.shape)
print(df1.info())
print("\nSummary:\n",df1.describe(include=['object']))

print(df1.isnull().sum())
#print(df1.value_counts())

print("\nFraud Class Counts:\n",(df1['is_fraud']).value_counts())

print("Fraud Rows:\n",df1[df1['is_fraud']==1].head())
print("Non-Fraud Rows:\n",df1[df1['is_fraud']==0].head())

print(pd.crosstab(df1['is_fraud'],df1["is_anomaly"],normalize=True))
print("Percentage:",df1["is_fraud"].value_counts(normalize=True))
print("Percentage:",df1["is_anomaly"].value_counts(normalize=True))
```

```python
main_data=pd.concat([df1,df2,df3],axis=0,ignore_index=True)

print(main_data.head())
print(main_data.shape)
print(main_data.info())
print("\nSummary:\n",main_data.describe(include=['object']))

print(main_data.isnull().sum())
#print(main_data.value_counts())

print("\nFraud Class Counts:\n",(main_data['is_fraud']).value_counts())

print("Fraud Rows:\n",main_data[main_data['is_fraud']==1].head())
print("Non-Fraud Rows:\n",main_data[main_data['is_fraud']==0].head())

print(pd.crosstab(main_data['is_fraud'],main_data["is_anomaly"],normalize=True))
print("Percentage:",main_data["is_fraud"].value_counts(normalize=True))
print("Percentage:",main_data["is_anomaly"].value_counts(normalize=True))
```

```python
main_data.to_parquet(r'..\data\training\vynfi\main_data.parquet',index=False)
```

```python
dataframe=pd.read_parquet(r'..\data\training\vynfi\main_data.parquet')

train_df,test_df=train_test_split(dataframe,test_size=0.2,random_state=42,stratify=dataframe['is_fraud'])

train_df.to_parquet(r'..\data\training\vynfi\train.parquet',index=False)
test_df.to_parquet(r'..\data\training\vynfi\test.parquet',index=False)
```
